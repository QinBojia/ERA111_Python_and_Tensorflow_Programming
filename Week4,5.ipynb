### Useful List Functions

```python
print(len(list1))
print(max(list1))
print(min(list1))
print(sum(list1))
print(sorted(list1))
print(list1)
print(any(list1)) # returns True if any element is True
print(all(list1)) # returns True if all elements are True
```

## List Approaches

```python
print(list1.index(8)) # parameter are the values to search for
print(list1.count(2))
list1.append(13)
print(list1)
list1.remove(13)
print(list1)
list1.pop()
print(list1)
list1.insert(1, 100) # parameters are the index and the value to insert
print(list1)
list1.sort()
print(list1)
list1.remove(100)
print(list1)
# list1.remove(100) ValueError: list.remove(x): x not in list
list1.clear()
```

## For Loop

```python
for i in [1, 2, 3, 4, 5]:
    print(i)
```

```python
for i in range(5):
    print(i)
for i in range(1, 10): # 10 not included
    print(i)
for i in range(1, 10, 3): print(i)
```

### An variation of for loop

```python
print([i for i in range(1, 11) if i % 2 == 0])
```

## While Loop

```python
x = 0
while x < 10:
   print(x, "Hello! I am inside while")
   x = x + 1
```

## Keywords in Loops

```python
for i in range(1, 10):
    if i == 1:
        continue
    elif i == 2:
        pass # doesn't do anything
    elif i == 3:
        break
    print(i)
```

```python
for i in range(1,10):
    for j in range(1,10):
        if j <= i: print(i * j, end=" ")
    print("\n")
```

```python
for i in range(1,4):
    for j in range(1,4):
        print(i*j, end=" ")
    print("\n")
```
