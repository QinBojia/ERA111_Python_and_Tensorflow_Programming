## Tuple
- Ordered
- Immutable
- Represented with ()
- Can be used as dictionary keys (lists cannot)

```python
t1 = (1, 2, 3)
t2 = ("apple", "banana")
t3 = (1, 2, "hello", 3.14)
t4 = () # empty tuple
t5 = (5,) # IMPORTANT: one-element tuple
t6 = (5,)
t7 = (5)  # This is an integer

print(t5==t6)
print(t5)
```

### Functions and Methods
| Operation              | Description                        |
|------------------------|------------------------------------|
| `x in s`               | True if element x is in sequence s |
| `x not in s`           | True if x is not in s              |
| `s1 + s2`              | Concatenation of two sequences     |
| `s * n`                | n copies of sequence s             |
| `s[i]`                 | ith element of sequence s          |
| `s[i:j]`               | Slice from index i to j−1          |
| `len(s)`               | Number of elements                 |
| `min(s)`               | Smallest element                   |
| `max(s)`               | Largest element                    |
| `sum(s)`               | Sum of numeric elements            |
| `for x in s:`          | Loop through items in s            |
| `<, >, <=, >=, ==, !=` | Sequence comparison                |

```python
a = (1, 2)
b = (1, 2, 3)
c = (1, 1, 0)

print(a<b)
print(a>c)

print(2*a)
print(a+a)
```

## Set
- Unordered
- Mutable
- No duplicates allowed
- Not particular order
- iterable

```python
s1 = {1, 2, 3}
s2 = {"apple", "banana"}
s3 = set() # empty set
s4 = set([1, 4, 9]) # {1, 4, 9}
s5 = set("xyz") # {'z', 'y', 'x'}
s6 = set(range(2,5)) # {2, 3, 4}
```

```python
a = {} # an empty dictionary

print(type(a))
```

```python
s = {1, 2, 3}
s.add(4)
s.remove(2) # error if not exists
s.discard(10) # NO error if not exists
s.update([5, 6])
print(s)
```

### Set Operations

```python
A = {1, 2, 3}
B = {3, 4, 5}
print(A | B) # Union → {1,2,3,4,5}
print(A & B) # Intersection → {3}
print(A - B) # Difference → {1,2}
print(A ^ B) # Symmetric difference → {1,2,4,5}
```

## Dictionary


```python
student = {
"name": "Alice",
"age": 20,
"major": "Marine Biotech"} # can be defined inter-lines
print(student)

info = {
"name": "Alice",
1001: "student id",
True: "status",
(3, 5): "Treasure"
}
print(info)
# values can be anything
# dict is iterable
```

###  Creating/Accessing dict

```python
d1 = {"name": "Bob", "age": 25}
d2 = dict(id=123, score=90)
d3 = {} # empty dictionary
a = dict()

student = {"name": "Alice", "age": 20}
print(student["name"])
print(student.get("age"))
print(student.get("address"))
print(student.get("address", "assigned output"))
```

### Adding/Updating/Removing Items

```python
# Adding/Updating

student = {"name": "Alice", "age": 20, "major":"Marine Biotech"}
student["age"] = 21
student["email"] = "alice@gmail.com"
student.update({"major": "CS"})
student.update(gender="Unknown")
print(student)

# Removing
student = {"name": "Alice", "age": 20, "major": "Marine Biotech"}
del student["name"]
student.pop("age")
print(student)
student.clear() # remove all items
```

Looping Through Dictionaries

```python
student = {"name": "Bob", "age": 22}
for key in student:
    print(key)
for value in student.values():
    print(value)
for key, value in student.items():
    print(key, ":", value)
```
