### Arguments

```python
def show_info(**data):
    for key, value in data.items():
        print(key, "=", value)
show_info(name="Alice", age=21, country="Malaysia")
dic = {"name":"Alice", "age":21, "country":"Malaysia"}
show_info(**dic)
# name="Alice" is called a Keyword Argument
```

```python
def calculate_price(item, *, quantity, price_per_unit): # All parameters following a bare asterisk parameters must be passed as keyword arguments
    total = quantity * price_per_unit
    print(f"Total for {item}: ${total}")

calculate_price("Apples", quantity=5, price_per_unit=1.5)
# Output: Total for Apples: $7.5
# calculate_price("Oranges", 3, 2.0) will raise a TypeError
```
